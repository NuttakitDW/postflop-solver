# Inspect Training Data (100K)

```python
import numpy as np
import pandas as pd

DATA_DIR = "../data/training_data_100k"

inputs = np.load(f"{DATA_DIR}/inputs.npy")
targets = np.load(f"{DATA_DIR}/targets.npy")

print(f"inputs shape: {inputs.shape}, dtype: {inputs.dtype}")
print(f"targets shape: {targets.shape}, dtype: {targets.dtype}")
```

## Input Features Layout

| Columns | Description |
|---------|-------------|
| 0–14    | Board features (15 one-hot: 4 flop cards×suit encoding + turn card) |
| 15–1014 | OOP bucketed range (1000 buckets) |
| 1015–2014 | IP bucketed range (1000 buckets) |

```python
# Build column names
board_cols = [f"board_{i}" for i in range(15)]
oop_cols = [f"oop_bucket_{i}" for i in range(1000)]
ip_cols = [f"ip_bucket_{i}" for i in range(1000)]

input_df = pd.DataFrame(inputs, columns=board_cols + oop_cols + ip_cols)
input_df.head()
```

```python
# Targets: CFVs for OOP (1000) + IP (1000)
oop_cfv_cols = [f"cfv_oop_{i}" for i in range(1000)]
ip_cfv_cols = [f"cfv_ip_{i}" for i in range(1000)]

target_df = pd.DataFrame(targets, columns=oop_cfv_cols + ip_cfv_cols)
target_df.head()
```

```python
# Board features (first 15 columns)
print("Board features — first 5 samples:")
input_df[board_cols].head()
```

```python
# Summary statistics
print("=== Input Stats ===")
print(f"Board features range: [{inputs[:, :15].min():.4f}, {inputs[:, :15].max():.4f}]")
print(f"OOP range sum (mean): {inputs[:, 15:1015].sum(axis=1).mean():.4f}")
print(f"IP  range sum (mean): {inputs[:, 1015:].sum(axis=1).mean():.4f}")
print()
print("=== Target Stats ===")
print(f"OOP CFV range: [{targets[:, :1000].min():.4f}, {targets[:, :1000].max():.4f}]")
print(f"IP  CFV range: [{targets[:, 1000:].min():.4f}, {targets[:, 1000:].max():.4f}]")
print(f"OOP CFV mean:  {targets[:, :1000].mean():.4f}")
print(f"IP  CFV mean:  {targets[:, 1000:].mean():.4f}")
```

```python
# Unique boards count
board_features = inputs[:, :15]
unique_boards = np.unique(board_features, axis=0)
print(f"Unique boards: {len(unique_boards)} out of {len(inputs)} samples")
print(f"Avg samples per board: {len(inputs) / len(unique_boards):.1f}")
```
